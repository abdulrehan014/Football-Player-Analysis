### Data Cleaning and Type Conversion

Based on the `df.info()` output, several columns that should be numerical are currently `object` type. This is due to non-numeric characters (e.g., '€', 'K', 'M', '+', '-', 'cm', 'kg') or special characters like '+'. The 'Unnamed: 64' column is also entirely empty. Let's clean these columns by removing non-numeric characters, converting them to numeric types, and dropping the 'Unnamed: 64' column.

````python
import re
import pandas as pd
import numpy as np # Import numpy for np.nan

# Drop the 'Unnamed: 64' column as it's completely empty
if 'Unnamed: 64' in df.columns:
    df = df.drop(columns=['Unnamed: 64'])
    print("Dropped 'Unnamed: 64' column.")

# Columns that may contain '€', 'K', 'M' multipliers (e.g., Value, Wage, Release clause)
money_cols = ['Value', 'Wage', 'Release clause']
for col in money_cols:
    if col in df.columns and df[col].dtype == 'object':
        temp_series = df[col].astype(str).str.replace('€', '', regex=False).str.strip()
        temp_series = temp_series.str.replace('K', '*1e3', regex=False)
        temp_series = temp_series.str.replace('M', '*1e6', regex=False)
        # Evaluate expressions and convert to numeric, coercing errors
        df[col] = temp_series.apply(lambda x: eval(x) if isinstance(x, str) and ('*' in x) else x)
        df[col] = pd.to_numeric(df[col], errors='coerce')
        print(f"Cleaned and converted '{col}' to numeric (money format).")

# Drop 'Height' and 'Weight' columns as they consistently fail to parse into meaningful numerical data
if 'Height' in df.columns:
    df = df.drop(columns=['Height'])
    print("Dropped 'Height' column as it contained unparseable data.")
if 'Weight' in df.columns:
    df = df.drop(columns=['Weight'])
    print("Dropped 'Weight' column as it contained unparseable data.")

# All other skill/rating columns that are currently 'object' type
# These should be simple numerical values, possibly with a '+' for potential,
# so we extract the base number and ensure only digits are present.
rating_skill_cols = [
    'Overall rating', 'Potential',
    'Crossing', 'Finishing', 'Heading accuracy', 'Short passing', 'Volleys',
    'Dribbling', 'Curve', 'FK Accuracy', 'Long passing', 'Ball control',
    'Acceleration', 'Sprint speed', 'Agility', 'Reactions', 'Balance',
    'Shot power', 'Jumping', 'Stamina', 'Strength', 'Long shots',
    'Aggression', 'Interceptions', 'Att. Position', 'Vision', 'Penalties', 'Composure',
    'Defensive awareness', 'Standing tackle', 'Sliding tackle',
    'GK Diving', 'GK Handling', 'GK Kicking', 'GK Positioning', 'GK Reflexes'
]

for col in rating_skill_cols:
    # This condition ensures we only process columns that are currently objects and in our list
    if col in df.columns and df[col].dtype == 'object':
        # Split by '+' and take the first part (base rating)
        temp_series = df[col].astype(str).str.split('+').str[0].str.strip()
        # Remove any non-numeric characters before conversion to handle anomalies like '861XYZ'
        temp_series = temp_series.str.replace(r'[^\d.]', '', regex=True)
        # Convert to numeric, coercing errors, and explicitly cast to float64
        df[col] = pd.to_numeric(temp_series, errors='coerce').astype(float)
        print(f"Cleaned and converted '{col}' to numeric (rating/skill format).")

# --- Additional Cleaning for 'Overall rating' and 'Potential' ---
# Player ratings are typically 0-99. Replace values > 100 with np.nan.
for col in ['Overall rating', 'Potential']:
    if col in df.columns:
        # Ensure the column is numeric after previous processing, coercing any remaining errors
        df[col] = pd.to_numeric(df[col], errors='coerce')

        initial_nan_count = df[col].isnull().sum()
        # Replace values greater than 100 with np.nan using .loc for direct assignment
        df.loc[df[col] > 100, col] = np.nan
        new_nan_count = df[col].isnull().sum()
        if new_nan_count > initial_nan_count:
            print(f"Cleaned '{col}': Replaced {new_nan_count - initial_nan_count} values > 100 with NaN.")

# Re-check the DataFrame info after cleaning and conversion
print('\nDataFrame Info after cleaning:')
df.info()

# Display the first few rows to see the cleaned data
print('\nFirst 5 rows after cleaning:')
display(df.head())
````

### Random Sampling Techniques

Now, let's explore different random sampling techniques and observe their impact on sample characteristics.

````python
# 1. Simple Random Sampling
sample_size = 100 # Let's take a sample of 100 players
srs_sample = df.sample(n=sample_size, random_state=42)

print(f"\n--- Simple Random Sample (n={sample_size}) ---")
display(srs_sample.head())
print(f"Mean 'Overall rating' in SRS: {srs_sample['Overall rating'].mean():.2f}")
print(f"Mean 'Potential' in SRS: {srs_sample['Potential'].mean():.2f}")
````

The simple random sample provides a basic, unbiased selection of data points. We can compare its statistics (e.g., mean 'Overall rating' and 'Potential') to those of the full dataset to see how well a small sample represents the population. In this case, the sample means are quite close to the population means, which is a good indicator, though this can vary with sample size and population distribution.

### 2. Stratified Sampling

Stratified sampling ensures that subgroups (strata) of a given population are represented in the sample proportionally. Let's stratify by 'foot' preference to ensure we get a proportional representation of left-footed and right-footed players.

````python
# Stratified Sampling by 'foot' preference
# We'll aim for a total sample size of 100, maintaining the original proportions of 'Left' and 'Right' foot.

# Calculate proportions of 'foot' in the original dataset
foot_proportions = df['foot'].value_counts(normalize=True)

# Determine sample sizes for each stratum
stratified_sample_size = 100
sample_per_stratum = (foot_proportions * stratified_sample_size).round().astype(int)

# Adjust for potential rounding issues (ensure total matches desired sample_size)
if sample_per_stratum.sum() != stratified_sample_size:
    diff = stratified_sample_size - sample_per_stratum.sum()
    # Add/subtract the difference from the largest stratum
    largest_stratum_idx = sample_per_stratum.idxmax()
    sample_per_stratum[largest_stratum_idx] += diff

print(f"Desired sample per stratum: {sample_per_stratum.to_dict()}")

# Perform stratified sampling
stratified_sample_list = []
for foot_type, num_samples in sample_per_stratum.items():
    stratum_df = df[df['foot'] == foot_type]
    stratified_sample_list.append(stratum_df.sample(n=num_samples, random_state=42))

stratified_sample = pd.concat(stratified_sample_list)

print(f"\n--- Stratified Sample (n={stratified_sample_size}) by 'foot' ---")
display(stratified_sample.head())
print(f"Distribution of 'foot' in Stratified Sample:\n{stratified_sample['foot'].value_counts(normalize=True).mul(100).round(2)}\n")
print(f"Mean 'Overall rating' in Stratified Sample: {stratified_sample['Overall rating'].mean():.2f}")
print(f"Mean 'Potential' in Stratified Sample: {stratified_sample['Potential'].mean():.2f}")
````

### 3. Systematic Sampling

Systematic sampling involves selecting elements from an ordered sampling frame at regular intervals. Let's sort the DataFrame by 'Overall rating' and then pick every k-th element.

````python
# Systematic Sampling
sys_sample_size = 100

# Sort the DataFrame by a relevant column (e.g., 'Overall rating')
df_sorted = df.sort_values(by='Overall rating').reset_index(drop=True)

# Calculate the sampling interval (k)
k = len(df_sorted) // sys_sample_size

# Select elements at regular intervals
systematic_sample = df_sorted.iloc[::k].head(sys_sample_size)

print(f"\n--- Systematic Sample (n={sys_sample_size}) ---")
display(systematic_sample.head())
print(f"Mean 'Overall rating' in Systematic Sample: {systematic_sample['Overall rating'].mean():.2f}")
print(f"Mean 'Potential' in Systematic Sample: {systematic_sample['Potential'].mean():.2f}")
````

### 4. Applying the Central Limit Theorem (CLT)

The Central Limit Theorem states that the distribution of sample means of a sufficiently large number of samples from any population will be approximately normal, regardless of the population's original distribution. Let's demonstrate this using the 'Overall rating' feature.

````python
import numpy as np

# Define parameters for CLT demonstration
population = df['Overall rating'].dropna() # Use 'Overall rating' as the population
sample_sizes = [10, 30, 100] # Different sample sizes to observe the effect
num_samples = 1000 # Number of samples to draw for each sample size

plt.figure(figsize=(15, 5))

for i, sample_size in enumerate(sample_sizes):
    sample_means = []
    for _ in range(num_samples):
        # Draw a random sample with replacement and calculate its mean
        sample = population.sample(n=sample_size, replace=True, random_state=i*42) # Unique random_state for each size
        sample_means.append(sample.mean())

    plt.subplot(1, len(sample_sizes), i + 1)
    sns.histplot(sample_means, kde=True, bins=30, color='lightgreen')
    plt.title(f'Distribution of Sample Means (n={sample_size})')
    plt.xlabel('Sample Mean of Overall Rating')
    plt.ylabel('Frequency / Density')
    # Plot population mean for reference
    plt.axvline(population.mean(), color='red', linestyle='dashed', linewidth=1,
                label=f'Population Mean: {population.mean():.2f}')
    plt.legend()

plt.tight_layout()
plt.show()
````

As observed from the plots, even though the original 'Overall rating' distribution might not be perfectly normal (though it's somewhat bell-shaped), the distribution of sample means tends to become more normal and centered around the population mean as the sample size (`n`) increases. This beautifully illustrates the Central Limit Theorem.

```markdown
## Lab 5: Confidence Intervals and Hypothesis Testing (Z-test and t-test)

In this section, we will:
1. Estimate the confidence intervals for key population parameters (e.g., the mean `Overall rating`).
2. Perform a one-sample Z-test (assuming known population variance or large sample size).
3. Perform a one-sample t-test (for situations where the population variance is unknown).
```

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# Fallback data loading and cleaning if 'df' is not in scope
if 'df' not in globals():
    print("Loading and cleaning dataset...")
    df = pd.read_csv('players.csv')
    if 'Unnamed: 64' in df.columns:
        df = df.drop(columns=['Unnamed: 64'])
    df['Overall rating'] = pd.to_numeric(df['Overall rating'].astype(str).str.split('+').str[0].str.replace(r'[^\d.]', '', regex=True), errors='coerce')
    df.loc[df['Overall rating'] > 100, 'Overall rating'] = np.nan

# Clean up any NaN values for the analysis
overall_ratings = df['Overall rating'].dropna()

# --- 1. Confidence Interval Estimation ---
mean_val = overall_ratings.mean()
std_err = stats.sem(overall_ratings)
confidence_level = 0.95

# 95% Confidence Interval for the mean overall rating
ci = stats.t.interval(confidence_level, len(overall_ratings)-1, loc=mean_val, scale=std_err)
print(f"95% Confidence Interval for the mean 'Overall rating': {ci[0]:.4f} to {ci[1]:.4f}")

# --- 2. One-Sample Z-test ---
null_mean = 73.0
population_std = overall_ratings.std()
z_stat = (mean_val - null_mean) / (population_std / np.sqrt(len(overall_ratings)))
p_val_z = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"\nOne-Sample Z-test (H0: Mean = {null_mean}):")
print(f"  Z-statistic: {z_stat:.4f}")
print(f"  p-value: {p_val_z:.4e}")

# --- 3. One-Sample t-test ---
t_stat, p_val_t = stats.ttest_1samp(overall_ratings, null_mean)

print(f"\nOne-Sample t-test (H0: Mean = {null_mean}):")
print(f"  t-statistic: {t_stat:.4f}")
print(f"  p-value: {p_val_t:.4e}")
````

## Advanced Analytics: Correlation, Covariance, and Machine Learning

In this section, we will:
1. Re-analyze Correlation and Covariance matrices for key target features.
2. Split our cleaned player dataset into Train and Test subsets ($80/20$ split).
3. Train, optimize, and evaluate five predictive models to predict a player's market `Value`:
   - **Multiple Linear Regression**
   - **Ridge Regression**
   - **Lasso Regression**
   - **Decision Tree Regressor**
   - **Random Forest Regressor**

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 1. Fallback Loading and Preprocessing Checks
if 'df' not in globals() or 'Value' not in df.columns or df['Value'].dtype == object:
    print("Preparing dataset for machine learning...")
    df = pd.read_csv('players.csv')
    if 'Unnamed: 64' in df.columns:
        df = df.drop(columns=['Unnamed: 64'])

    for col in ['Overall rating', 'Potential', 'Age']:
        if col in df.columns:
            df[col] = df[col].astype(str).str.split('+').str[0].str.strip()
            df[col] = pd.to_numeric(df[col].str.replace(r'[^\d.]', '', regex=True), errors='coerce')
            df.loc[df[col] > 100, col] = np.nan

    for col in ['Value', 'Wage']:
        if col in df.columns:
            temp = df[col].astype(str).str.replace('€', '', regex=False).str.strip()
            temp = temp.str.replace('M', 'e6', regex=False).str.replace('K', 'e3', regex=False)
            df[col] = pd.to_numeric(temp, errors='coerce')

# Prepare subset with selected numeric features
features = ['Overall rating', 'Potential', 'Age', 'Wage']
target = 'Value'

ml_df = df[features + [target]].dropna().copy()
for col in ml_df.columns:
    ml_df[col] = ml_df[col].astype(float)

# --- 2. Covariance and Correlation Analysis ---
print("=== Covariance Matrix ===")
display(ml_df.cov())

print("\n=== Correlation Matrix (Pearson) ===")
display(ml_df.corr())

# --- 3. Train-Test Split ---
X_ml = ml_df[features]
y_ml = ml_df[target]

X_train, X_test, y_train, y_test = train_test_split(X_ml, y_ml, test_size=0.2, random_state=42)
print(f"\nTrain-Test Split Completed:")
print(f"  Training set shape: {X_train.shape}")
print(f"  Testing set shape:  {X_test.shape}")
````

````python
# --- 4. Predictive Modeling and Evaluation ---
models = {
    "Multiple Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=1.0, max_iter=10000),
    "Decision Tree Regressor": DecisionTreeRegressor(max_depth=5, random_state=42),
    "Random Forest Regressor": RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
}

results = {}

print("=== Model Performance Summary ===\n")
for name, model in models.items():
    # Train
    model.fit(X_train, y_train)
    # Predict
    y_pred = model.predict(X_test)
    # Evaluate
    r2 = r2_score(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    results[name] = {"R2": r2, "RMSE": rmse, "Predictions": y_pred}
    print(f"{name}:")
    print(f"  R² Score: {r2:.4f}")
    print(f"  RMSE:     €{rmse:,.2f}\n")

# --- 5. Visualization of Model Predictions ---
plt.figure(figsize=(12, 6))
plt.scatter(y_test, results["Random Forest Regressor"]["Predictions"], alpha=0.5, label='Random Forest (Best Model)', color='darkgreen')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Perfect Fit')
plt.xlabel('Actual Player Value (€)')
plt.ylabel('Predicted Player Value (€)')
plt.title('Actual vs. Predicted Player Value')
plt.legend()
plt.show()
````

### Comprehensive Explanation of Regression Models

In our analytics workflow, we implemented five different types of regression algorithms to predict player market `Value`. Here is a detailed breakdown of how each model functions, its objective, and its advantages/disadvantages.

---

#### 1. Multiple Linear Regression
*   **Concept**: Multiple Linear Regression finds the linear relationship between two or more independent variables (predictors like *Overall rating*, *Potential*, *Age*, *Wage*) and a single dependent target variable (*Value*).
*   **Mathematical Formula**:
    $$\hat{y} = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n$$
    Where $\beta_0$ is the intercept, and $\beta_1, \beta_2, \dots, \beta_n$ are the regression coefficients representing the change in target per unit change in predictor.
*   **Advantages**: Very easy to interpret, fast to train, and acts as an excellent baseline model.
*   **Disadvantages**: Assumes a strictly linear relationship. It is highly sensitive to outliers and multicollinearity (when independent variables are highly correlated).

---

#### 2. Ridge Regression ($L_2$ Regularization)
*   **Concept**: Ridge regression is an extension of linear regression that introduces a regularization penalty to avoid **overfitting** and handle multicollinearity. It penalizes large coefficients by adding their squared values to the loss function.
*   **Mathematical Objective**:
    $$\text{Minimize } \sum (y - \hat{y})^2 + \alpha \sum_{j=1}^{n} \beta_j^2$$
    The hyperparameter $\alpha$ (alpha) controls the strength of the penalty.
*   **Advantages**: Shrinks the coefficients close to zero, making the model much more stable when features are correlated.
*   **Disadvantages**: It shrinks coefficients but never forces them to exactly zero; thus, it does not perform feature selection.

---

#### 3. Lasso Regression ($L_1$ Regularization)
*   **Concept**: Lasso (Least Absolute Shrinkage and Selection Operator) is another regularized linear model. Instead of squaring the coefficients, it penalizes the absolute values of the coefficients.
*   **Mathematical Objective**:
    $$\text{Minimize } \sum (y - \hat{y})^2 + \alpha \sum_{j=1}^{n} |\beta_j|$$
*   **Advantages**: Lasso can shrink coefficients *completely* to zero. This makes it highly useful for **automatic feature selection**.
*   **Disadvantages**: If there are highly correlated variables, Lasso arbitrarily selects only one of them and sets the others to zero.

---

#### 4. Decision Tree Regressor
*   **Concept**: A non-linear algorithm that splits the dataset into smaller, homogeneous subsets based on decision boundaries (e.g., *Is Overall rating > 75?*). It builds a tree-like structure where each leaf node represents a numerical prediction (usually the mean of training targets in that node).
*   **Advantages**: Can capture complex, non-linear relationships and interactions without needing feature scaling or assuming linear distributions.
*   **Disadvantages**: Highly prone to **overfitting** (growing too deep and memorizing training data noise), which can lead to unstable predictions on new data.

---

#### 5. Random Forest Regressor
*   **Concept**: An ensemble method that builds an array (forest) of independent **Decision Trees** trained on random subsets of the data (bootstrapping) and random subsets of features. The final prediction is the **average** of the predictions from all individual trees.
*   **Advantages**: Reduces overfitting dramatically compared to a single decision tree, yields high predictive power, handles non-linear patterns, and is robust to outliers.
*   **Disadvantages**: More computationally expensive, slower to train, and behaves as a "black-box" model, which makes it harder to interpret than simple linear regressions.
